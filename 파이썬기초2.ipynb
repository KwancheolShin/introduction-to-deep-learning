#6. 문자열 Formatting 출력 방법

f"{variable:format_specifier}" 형태로 입력한다.


```python
name = "Alice"; age = 25 #변수 지정
print(f"이름:{name}, 나이:{age}")
```

```python
x = 10
y = 3

print(f"x + y = {x + y}")
print(f"x * y = {x * y}")
```

실수의 소수점 자리 지정할 수 있다. 수치 계산에서는 중요하다. 너무 긴 숫자는 자릿수를 지정해서 통일감을 준다.

```python
pi = 3.14159265358979

print(f"pi = {pi}")
print(f"pi = {pi:.2f}") #소숫점 2자리
print(f"pi = {pi:.4f}") #소숫점 4자, f는 float의 f이다
```

전체 출력 폭(width) 지정

```python
x = 3.14159

print(f"{x:30.2f}") #10-> 전체 출력 공간을 10칸 확보; .2-> 소수점 아래 2자리, f->fixed-point 실수형식
```

```python
name = "Alice"
score = 93.456

print(f"{name:10s} {score:8.2f}") #표를 출력할때 유용하다.
```

```python
word = "Python"

print(f"|{word:<10}|")   # < 왼쪽 정렬
print(f"|{word:^10}|")   # ^ 가운데 정렬
print(f"|{word:10>}|")   # > 오른쪽 정렬
```

```python
n = 1234567
print(f"{n:,}") # 정수 포맷팅. :,는 큰 숫자를 쉼표로 구분.
print(f"{n:010d}") # 10칸을 만들고 빈자리는 0으로 채운다.
```

```python
price = 1250000
print(f"Price: {price:,} won")
```

파이썬에서 for문은 같은 작업을 여러 번 반복(iteration)할 때 사용하는 기본적인 반복문이다. iteratable object에서 하나씩 순회하면서 값을 가져온다.
주어진 시퀀스(리스트, 튜플, 문자열, range 등)의 요소를 하나씩 가져와 반복을 수행한다.

list 순회

```python
fruits = ["사과","바나나","포도"]
for fruit in fruits: #리스트 순회
  print(fruit)
```

```python
for i in [1,2,3]:
  print(i)
```

```python
for i in range(3): #range() 함수로 for문 실행
  print(i)
```

```python
for i in range(1,4): #range의 범위를 지정할 수도 있다. range(start,stop)
  print(i)
```

```python
for i in range(2,10,2): # 간격(step)을 조정할 수도 있다. 1부터 10까지 2씩 증가, range(start,stop,step)
  print(i)
```

zip()을 사용하면 여러 리스트를 한 번에 순회할 수 있다. zip은 여러 리스트의 요소를 튜플로 묶어 반환한다.

#9. while 문
while 문은 특정 **조건이 참(True)인 동안에 계속 실행되는 반복문**이다. 조건이 거짓(False)이 되면 반복이 종료된다.
while문도 break와 continue를 사용할 수 있다.

```python
i = 1
while i<=10:
  print(i)
  i+=2
```

```python
i = 1
while i<5:
  if i==3: #i==3을 만나면 break를 만나 반복문이 종료된다.
    break
  print(i)
  i+=1
```

```python
i = 0
while i<5:
  i+=1
  if i ==3:
    continue #i가 3이면 출력하지 않고 건너뛴다.
  print(i)
```

##12.3. 가변 인자(\*args)

몇 개의 인자가 전달될지 미리 정해져 있지 않은 경우 가변 인자(variable-length arguments)를 사용할 수 있다.
\*args를 사용해 임의 개의 인자를 튜플 형태로 받는다.


```python
def sum_all(*args):
  return sum(args)

print(sum_all(1,2,3))
print(sum_all(4,5,6,7,8))
```

```python
def student(name, *scores): #일반인자와 가변인자를 함께 사용
    print("Name:", name)
    print("Scores:", scores)

student("Kim", 90, 85, 95) #첫번째 인자는 name이 받고, 나머지 위치들의 인자는 scores가 모두 받아 튜플로 묶는다.
```

키워드가변인자(**kwargs)
keyword arguments를 받을 때 사용한다.

\**kwargs를 사용해 임의 개의 매개변수와 인자를 받을수 있다. kwargs는 튜플이 아니라 딕셔너리 형태로 저장된다.
kwargs.items()로 키-값을 가져올 수 있다.

```python
def introduce(**kwargs):
  for key, value in kwargs.items():
    print(f"{key}: {value}")

introduce(name="Alice", age=25, city="New York")
```

```python
def show_info(title, *args, **kwargs): #*args와 **kwargs를 함께 사용할수 있다
  print(f"제목: {title}")

  print("위치 인자:", args) #튜플
  for key, value in kwargs.items():
    print(f"{key}: {value}")

show_info("학생정보", "서울","대학생", 이름="Kwancheol", 나이=20, 전공="Mathematics")

#*args는 위치 인자이고 자료형은 tuple이다. **kargs는 키워드 인자이고 dict로 받는다
```

#13. 클래스
클래스는 연관된 데이터(변수)와 기능(함수)을 하나로 묶어서 관리하는 구조이다. 이를 통해 **객체지향 프로그래밍(Object-Oriented Programming)**을 구현할 수 있다. 객체지향 프로그래밍은 프로그램을 **객체(object)**들의 모임으로 구성하는 프로그래밍 방식이다.

각 객체는 보통
**데이터(data) + 그 데이터를 처리하는 기능(function)**
을 함께 가지고 있다. 파이썬에서는 이를 **클래스(class)**를 이용해 구현한다.

학생 정보를 저장하고 관리해야 한다고 가정하고 클래스를 활용하는 예시를 들어보자.

```python
S1 = {"name": "Alice", "age": 20, "major": "Mathematics"}
S2 = {"name": "Bob", "age": 21, "major": "Computer Science"}

print(S1["name"])
```

클래스를 활용하면 학생정보를 담을 수 있는 클래스를 만들어 학생 개개인을 클래스의 하나의 객체(Object)로 다룰수 있다.

```python
class Student:
  #클래스 변수(class variables)을 지정, 모든 Student 객체가 공유한다
  school = "Korea University"
  count = 0

  def __init__(self,name,age,student_id,major): #클래스의 속성을 초기화(정의)한다.
    self.name = name
    self.age = age
    self.student_id = student_id
    self.major = major
    Student.count += 1 #Student클래스로 객체를 하나 만들때 마다 count를 하나씩 늘림

  @classmethod #클래스 메소드를 지정, 호출은 Class.method()
  def show_info(cls): #cls는 @classmethod 안에서 Student 클래스 자체를 가리키는 이름
    print(cls.school) #cls.school는 사실상 Student.school과 같은 의미이지
    print(cls.count)

  #instance method는 각 개체의 속성변수 self를 이용해서 정의한다. 호출은 obj.method()
  def introduce(self): #클래스 안에 클래스의 메서드(method)를 만들어 사용할 수 있다. method를 함수(function)라고도 한다.
    #self란 객체 자기 자신을 가르키는 속성변수로, instance method의 첫번째 인자는 self로 둔다.
    print(f"안녕하세요! 저는 {self.name}입니다. {self.major} 전공을 공부하고 있습니다.")


s1 = Student("김민경",20,"2025001","Computer Science") #클래스로 생성된 각각을 '객체' 또는 '인스턴스'라고 한다.
s2 = Student("이수지",20,"2025002","Mathematics")

print(s1.name) #[객체명.속성] 으로 인스턴스 변수에 접근할 수 있다.
Student.show_info() #Student라는 클래스의 클래스메서드를 실행한다.
s1.introduce() #student1의 introduce()함수를 실행한다.
```

```python
#학생을 여러 명 등록하고 관리하는 class를 만들어보자.
class StudentManager:
  def __init__(self):
    self.students = [] # 학생 객체를 저장할 빈 리스트를 만든다


  def add_student(self,student):
    self.students.append(student)
    # 리스트는 append함수를 이용해서 값을 추가할 수 있다. append를 이용해서 학생 객체를 리스트에 추가한다.

  def show_students(self):
    print("학생 목록:")
    for student in self.students:
      print(f"이름: {student.name}, 나이: {student.age}, 학번: {student.student_id}, 전공: {student.major}")

manager = StudentManager()
manager.add_student(s1) #위에서 Student 클래스로 생성한 Student 클래스의 객체 student1을 StudentManager로 생성한 객체 manager의 속성 students에 저장.
manager.add_student(s2)

manager.show_students()
```

```python
class Student:
    school = "Korea University"

    def __init__(self, name):
        self.name = name

    # instance method
    def introduce(self):
        print(f"I am {self.name}.")

    # class method
    @classmethod
    def show_school(cls):
        print(f"School: {cls.school}")

    # static method
    @staticmethod
    def is_adult(age):
        return age >= 18
```

```python
s1 = Student("Kim")

s1.introduce()  #s1이라는 학생의 특정 정보가 필요
Student.show_school() # Student 클래스 전체의 정보가 필요
print(Student.is_adult(20)) # Student 객체도, Student 클래스 정보도 필요없음. 단지 Student와 관련된 기능
```
