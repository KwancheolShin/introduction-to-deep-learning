```python
print(a[0])
print(a[1])
print(a[2])
print(a[3])
print(a[-1])
print(a[-2])
print(a[0:2]) #0번 인덱스부터, 인덱스 2 직전까지 슬라이싱 출력
print(a[2:4]) #2번 인덱스부터, 인덱스 4 직전까지, 2,3인덱스 출력
print(a[:3]) #처음부터 인덱스 3 직전까지, 인덱스 0,1,2 이 출력
print(len(a))

```

##3.2 딕셔너러dictionary
리스트는 인덱스 번호로 0,1,2, ... 순으로 값을 지정하는 반면, 딕셔너리dictionary는 키key와 값value을 한 쌍으로 저장한다. 즉, 영한사전처럼 단어와 그 의미를 짝지어 저장한다. 중괄호{}로 정의하며 키와 값은 콜론:으로 연결한다.
키를 이용하여 접근해서 값을 호출하고 변경하고 추가하고 삭제할 수 있다.

```python
kwancheol = {"이름":"신관철", "키":195, "몸무게":90, "나이":20, "학교":"고려대학교"}
```

```python
print(kwancheol)
print(type(kwancheol))
print( "나의 나이는=", kwancheol["나이"],"살 입니다.")
```

```python
kwancheol["주소"] = "천당"
kwancheol["학번"] = 2024
print(kwancheol)
```

```python
for a, b in kwancheol.items(): #for 반복문, item은 key와 value를 모두 반환하는 명령어
  print(a,b)
```

##3.3 bool(불)
파이썬에는 bool(불 또는 불리언)이라는 자료형이 있다. bool 자료형은 True(참)와 False(거짓)라는 두 값 중 하나를 취한다. bool에는 and, or, not 연산자를 사용할 수 있다.

```python
hungry = True
sleep  = False
```

```python
print(type(hungry))
print(hungry)
print(int(hungry))
print(sleep)
print(int(sleep))
```

#4. 논리연산자
수치연산에서 +,-,*,/ 등이 있듯이 자료형에 따라 사용할 수 있는 연산자가 정해져 있다. and, or, not

```python
print(hungry and hungry)
print(hungry and sleep)
print(sleep and sleep)
```

```python
print(hungry or hungry)
print(sleep or sleep)
print(hungry or sleep)
print(not hungry) #부정 연산자
print(not sleep)
```

#5. 조건문
조건에 따라 달리 처리하려면 if/else 문을 사용한다. if 다음줄에는 공백을 두 칸 또는 탭(tab)으로 일정하게 둔다.


```python
if hungry: #if (조건문):
  print("I'm hungry")
```

```python
if sleep:
  print("I'm sleepy")
else:
  print("I'm not sleepy.")
```

```python
a = 5
if a == 5:
  print("a의 값은 5 입니다.")
else:
  print("a의 값은 5가")
```
