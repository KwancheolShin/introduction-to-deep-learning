```python
import numpy as np
import matplotlib.pyplot as plt


# ============================================================
# Function
# f(x) = x^4 + x^2
# ============================================================

f   = lambda x: 1/4*x**4 +1/3*x**3+ 1/2*x**2
df  = lambda x: x**3 + x**2 + x
ddf = lambda x: 3*x**2 + 2*x+ 1


# ============================================================
# Initial point
# ============================================================

xk = 0.5

# x-axis for plotting
x = np.linspace(-0.5, 1.3, 500)


# ============================================================
# Newton Method: 3 iterations
# ============================================================

for k in range(3):

    # --------------------------------------------------------
    # Newton step
    # --------------------------------------------------------

    x_next = xk - df(xk) / ddf(xk)


    # --------------------------------------------------------
    # Quadratic approximation at x_k
    #
    # q_k(x)
    # = f(x_k)
    # + f'(x_k)(x-x_k)
    # + 1/2 f''(x_k)(x-x_k)^2
    # --------------------------------------------------------

    q = (
        f(xk)
        + df(xk) * (x - xk)
        + 0.5 * ddf(xk) * (x - xk)**2
    )


    # ========================================================
    # Plot
    # ========================================================

    plt.figure(figsize=(8, 6))


    # Original function
    plt.plot(
        x, f(x),
        'k',
        linewidth=2,
        label=r'$f(x)=x^4+x^2$'
    )


    # Quadratic approximation
    plt.plot(
        x, q,
        'b--',
        linewidth=2,
        label=rf'$q_{k}(x)$'
    )


    # --------------------------------------------------------
    # Current point x_k
    # --------------------------------------------------------

    plt.plot(
        xk, f(xk),
        'ro',
        markersize=8,
        label=rf'$x_{k}$'
    )


    # --------------------------------------------------------
    # Minimum of quadratic approximation = x_{k+1}
    # --------------------------------------------------------

    q_min = (
        f(xk)
        + df(xk) * (x_next - xk)
        + 0.5 * ddf(xk) * (x_next - xk)**2
    )

    plt.plot(
        x_next, q_min,
        'bs',
        markersize=8,
        label=rf'$x_{k+1}$: minimum of $q_{k}$'
    )


    # Vertical line showing x_{k+1}
    plt.axvline(
        x_next,
        color='gray',
        linestyle=':',
        linewidth=1.5
    )


    # --------------------------------------------------------
    # Show Newton step on x-axis
    # --------------------------------------------------------

    plt.annotate(
        '',
        xy=(x_next, 0),
        xytext=(xk, 0),
        arrowprops=dict(
            arrowstyle='->',
            linewidth=2
        )
    )


    # Labels for x_k and x_{k+1}
    plt.text(
        xk,
        -0.08,
        rf'$x_{k}={xk:.3f}$',
        ha='center',
        fontsize=11
    )

    plt.text(
        x_next,
        -0.18,
        rf'$x_{k+1}={x_next:.3f}$',
        ha='center',
        fontsize=11
    )


    # --------------------------------------------------------
    # Plot settings
    # --------------------------------------------------------

    plt.axhline(0, color='gray', linewidth=0.8)

    plt.xlabel('x', fontsize=12)
    plt.ylabel('f(x)', fontsize=12)

    plt.title(
        rf"Newton's Method: Step {k+1}",
        fontsize=14
    )

    plt.xlim(-0.5, 1.3)
    plt.ylim(-0.3, 4.0)

    plt.grid(alpha=0.25)
    plt.legend()

    plt.show()


    # --------------------------------------------------------
    # Print numerical information
    # --------------------------------------------------------

    print(f"Step {k+1}")
    print(f"x_{k}     = {xk:.6f}")
    print(f"f'(x_{k})  = {df(xk):.6f}")
    print(f"f''(x_{k}) = {ddf(xk):.6f}")
    print(f"x_{k+1}   = {x_next:.6f}")
    print()


    # Update
    xk = x_next
```
